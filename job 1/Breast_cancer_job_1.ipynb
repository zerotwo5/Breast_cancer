```python
from matplotlib.backends.backend_pdf import PdfPages
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
from scipy.stats import ttest_ind, norm, binom, t as st_t
import scipy.stats as st
import io
import sys

# Open the PDF file
with PdfPages('math_ai_assessment_results.pdf') as pdf:
    # 1. DataFrame Descriptive Statistics
    fig, ax = plt.subplots(figsize=(12, 8))
    ax.axis('off')
    ax.set_title('DataFrame Descriptive Statistics', fontsize=16)
    table_text = df.describe().to_string()
    ax.text(0.01, 0.99, table_text, transform=ax.transAxes,
            fontsize=8, verticalalignment='top', horizontalalignment='left',
            fontfamily='monospace')
    pdf.savefig(fig)
    plt.close(fig)

    # 2. T-test result for x.radius_mean
    # Ensure y_numeric is created if not already in the session
    if 'y_numeric' not in df.columns:
        df['y_numeric'] = df['y'].map({'B': 0, 'M': 1})
    group1 = df[df['y_numeric']==0]['x.radius_mean']
    group2 = df[df['y_numeric']==1]['x.radius_mean']
    ttest_result = ttest_ind(group1, group2)

    fig, ax = plt.subplots(figsize=(12, 2))
    ax.axis('off')
    ax.set_title('T-test Result for x.radius_mean by y_numeric', fontsize=14)
    ttest_str = f"Statistic: {ttest_result.statistic:.4f}\nP-value: {ttest_result.pvalue:.4e}\nDegrees of Freedom: {ttest_result.df:.0f}"
    ax.text(0.01, 0.9, ttest_str, transform=ax.transAxes,
            fontsize=12, verticalalignment='top', horizontalalignment='left',
            fontfamily='monospace')
    pdf.savefig(fig)
    plt.close(fig)

    # 3. 95% Confidence Interval for x.radius_mean
    data_ci = df['x.radius_mean']
    ci = st.t.interval(0.95, len(data_ci)-1, loc=np.mean(data_ci), scale=st.sem(data_ci))

    fig, ax = plt.subplots(figsize=(12, 2))
    ax.axis('off')
    ax.set_title('95% Confidence Interval for x.radius_mean', fontsize=14)
    ci_str = f"Lower bound: {ci[0]:.4f}\nUpper bound: {ci[1]:.4f}"
    ax.text(0.01, 0.9, ci_str, transform=ax.transAxes,
            fontsize=12, verticalalignment='top', horizontalalignment='left',
            fontfamily='monospace')
    pdf.savefig(fig)
    plt.close(fig)

    # 4. Skewness and Kurtosis for x.radius_mean
    skew_val = df.skew(numeric_only=True)['x.radius_mean']
    kurt_val = df.kurt(numeric_only=True)['x.radius_mean']

    fig, ax = plt.subplots(figsize=(12, 2))
    ax.axis('off')
    ax.set_title('Skewness and Kurtosis for x.radius_mean', fontsize=14)
    skew_kurt_str = f"Skewness (x.radius_mean): {skew_val:.4f}\nKurtosis (x.radius_mean): {kurt_val:.4f}"
    ax.text(0.01, 0.9, skew_kurt_str, transform=ax.transAxes,
            fontsize=12, verticalalignment='top', horizontalalignment='left',
            fontfamily='monospace')
    pdf.savefig(fig)
    plt.close(fig)

    # 5. Crosstab: Y vs. x.radius_mean > 15
    crosstab_df = pd.crosstab(df['y'], df['x.radius_mean']>15, normalize='index')
    fig, ax = plt.subplots(figsize=(12, 4))
    ax.axis('off')
    ax.set_title('Crosstab: Y vs. x.radius_mean > 15 (Normalized by Index)', fontsize=14)
    crosstab_str = crosstab_df.to_string(float_format="%.4f")
    ax.text(0.01, 0.99, crosstab_str, transform=ax.transAxes,
            fontsize=10, verticalalignment='top', horizontalalignment='left',
            fontfamily='monospace')
    pdf.savefig(fig)
    plt.close(fig)

    # 6. Histogram of x.radius_mean
    fig = plt.figure(figsize=(10, 6))
    sns.histplot(df['x.radius_mean'], kde=True)
    plt.title('Histogram of x.radius_mean', fontsize=16)
    plt.xlabel('x.radius_mean', fontsize=12)
    plt.ylabel('Frequency', fontsize=12)
    pdf.savefig(fig)
    plt.close(fig)

    # 7. Normal and Binomial Distributions Plot
    fig = plt.figure(figsize=(10, 6))
    x_dist = np.linspace(0, 10, 100)
    plt.plot(x_dist, norm.pdf(x_dist, 5, 1), label='Normal Distribution')
    plt.plot(range(10), binom.pmf(range(10), 10, 0.5), label='Binomial Distribution')
    plt.legend(fontsize=12)
    plt.title('Normal and Binomial Distributions', fontsize=16)
    plt.xlabel('Value', fontsize=12)
    plt.ylabel('Probability/Density', fontsize=12)
    pdf.savefig(fig)
    plt.close(fig)

print("All outputs, graphs, and results saved to math_ai_assessment_results.pdf")
```
